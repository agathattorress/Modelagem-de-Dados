```python
#Classe Estante
class Estante:
  def __init__ (self, tipo, objetos):
    self.tipo = tipo
    self.objetos = objetos
    self.estante = {}
    self.listaobj()

  def listaobj(self):
    self.estante = {}

    for i, objeto in enumerate(self.objetos, start=1):
      self.estante[str(i)] = objeto

  def imprimirlist(self):
    print(f"Escolha um {self.tipo}:")
    print(self.estante)

  def get_objeto(self, op):
    return self.estante[op]
```

```python
#Classe Produto
class Produto:
    def __init__(self, nome, descricao, preco, quantidade, categoria):
        self.nome = nome
        self.descricao = descricao
        self.preco = preco
        self.quantidade = quantidade
        self.categoria = categoria

    def __repr__(self):
        return self.nome

    def exibir_dados(self):
        print("Nome: ", self.nome)
        print("Descrição: ", self.descricao)
        print("Preço: ", self.preco)
        print("Quantidade: ", self.quantidade)
        print("Categoria: ", self.categoria, "\n")
```

```python
#Classe Animal
class Animal:
    def __init__(self, nome, descricao, raca, idade, porte):
        self.nome = nome
        self.descricao = descricao
        self.raca = raca
        self.idade = idade
        self.porte = porte

    def __repr__(self):
        return self.nome

    def exibir_dados(self):
        print("Nome:", self.nome)
        print("Descrição:", self.descricao)
        print("Raça:", self.raca)
        print("Idade:", self.idade)
        print("Porte:", self.porte, "\n")
```

```python
#Classe Serviço
class Servico:
  def __init__(self, nome, descricao, preco, duracao, categoria):
    self.nome = nome
    self.descricao = descricao
    self.preco = preco
    self.duracao = duracao
    self.categoria = categoria

  def __repr__(self):
        return self.nome

  def exibir_dados(self):
        print("Nome: ", self.nome)
        print("Descrição: ", self.descricao)
        print("Preço: ", self.preco)
        print("Duração: ", self.duracao)
        print("Categoria: ", self.categoria, "\n")
```

```python
#Classe cliente
class Cliente:
  def __init__(self, nome, telefone, email, endereco):
    self.nome = nome
    self.telefone = telefone
    self.email = email
    self.endereco = endereco

  def cadastrar_cliente(self):
    self.nome = input ("Digite seu nome: ").strip()
    self.telefone = input ("Digite seu telefone: ").strip()
    self.email = input ("Digite seu email: ").strip()
    self.endereco = input ("Digite seu endereço: ").strip()

  def exibir_dados(self):
    print("nome:", self.nome)
    print("telefone:", self.telefone)
    print("email:" , self.email)
    print("endereco" , self.endereco)
```

```python
#Classe evento
class Evento:
  def __init__(self, data, hora, servico, cliente):
    self.data = data
    self.hora = hora
    self.servico = servico
    self.cliente = cliente

  def criar_evento(self, servico, cliente):
    self.data = input("Digite uma data: ").strip()
    self.hora = input ("Digite um horário: ").strip()
    self.servico = servico
    self.cliente = cliente
```

```python
#Classe agenda
class Agenda:
  def __init__(self):
    self.agenda = []

  def adicionar_na_agenda(self, evento):
    self.agenda.append(evento)

  def listar_eventos(self):
    for evento in self.agenda:
      print(evento)
```

```python
#Salvar agendamento no arquivo

def salvar_agenda(evento):
    with open("dados.txt", "a", encoding="utf-8") as arquivo:
        arquivo.write(
            f"{evento.data};"
            f"{evento.hora};"
            f"{evento.servico.nome};"
            f"{evento.cliente.nome}\n"
        )
```

```python
#Carregar dados
def carregar_dados():
    produtos = []
    animais = []
    servicos = []

    secao = ""

    with open("dados.txt", "r", encoding="utf-8") as arquivo:
        for linha in arquivo:
            linha = linha.strip()

            if not linha:
                continue

            if linha.startswith("["):
                secao = linha
                continue

            dados = linha.split(";")

            if secao == "[PRODUTOS]":
                produto = Produto(
                    dados[0],
                    dados[1],
                    dados[2],
                    dados[3],
                    dados[4]
                )
                produtos.append(produto)

            elif secao == "[ANIMAIS]":
                animal = Animal(
                    dados[0],
                    dados[1],
                    dados[2],
                    dados[3],
                    dados[4]
                )
                animais.append(animal)

            elif secao == "[SERVICOS]":
                servico = Servico(
                    dados[0],
                    dados[1],
                    dados[2],
                    dados[3],
                    dados[4],
                )
                servicos.append(servico)

    return produtos, animais, servicos
```

```python
#Criar estantes com os dados carregadosda
produtos, animais, servicos = carregar_dados()

lproduto = Estante("produto", produtos)
lanimal = Estante("animal", animais)
lservico = Estante("servico", servicos)
```

```python
#Main
agenda = Agenda()

while True:
  cabecalho()
  print("Bem vindo ao Pet Care \n O que deseja ver hoje?")
  print("[1]Produtos [2]Serviços [3]Animais")
  print("[0]Sair")

  op = input("Escolha uma opção: ").strip()

  match op:
    case "0":
      break

    case "1":
      lproduto.imprimirlist()

    case "2":
      lservico.imprimirlist()

    case "3":
      lanimal.imprimirlist()

    case _:
      print("Opção inválida!")
      continue

  opp = input("Digite o número escolhido: ").strip()

  match op:

    case "1":  # Produto
      produto = lproduto.get_objeto(opp)
      produto.exibir_dados()

      escolha = input("[1] Prosseguir  [2] Voltar ao menu principal: ").strip()

      if escolha == "1":
        compra()
      elif escolha == "2":
        continue

    case "3":  # Animal
      animal = lanimal.get_objeto(opp)
      animal.exibir_dados()

      escolha = input("[1] Prosseguir  [2] Voltar ao menu principal: ").strip()

      if escolha == "1":
        adocao()
      elif escolha == "2":
        continue

    case "2":  # Serviço
      servico = lservico.get_objeto(opp)
      servico.exibir_dados()

      escolha = input("[1] Prosseguir  [2] Voltar ao menu principal: ").strip()

      if escolha == "1":
        cliente = Cliente("", "", "", "")
        cliente.cadastrar_cliente()

        evento = Evento("", "", servico, cliente)
        evento.criar_evento(servico, cliente)

        agenda.adicionar_na_agenda(evento)
        salvar_agenda(evento)
        convite()

      elif escolha == "2":
        continue

footer()
```
